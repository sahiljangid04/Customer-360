## null value handling
### columns containg null = orders_raw, products_raw, sellers_raw

```python
#orders
print(orders.isna().sum())
print(orders.shape)


#the dates nan will be keep as it as it can be the orders which is cancel never went for dilivery 
#or didnt recived to customres so it will be keep as nan

```

```python
#products
print('Before')
print(products.shape)
print(products.isna().sum())
#product_category_name         623 - replace with unknown catagoery
#product_name_lenght           610 - impute with median
#product_description_lenght    610 - impute with median
#product_photos_qty            610 - impute with median
#product_weight_g                2 - impute with median
#product_length_cm               2 - impute with median
#product_height_cm               2 - impute with median
#product_width_cm                2 - impute with median
products['product_category_name'] = (products['product_category_name'].fillna('unknown'))

missing_cols = [
    'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]

for col in missing_cols:
    products[col] = (
        products
        .groupby('product_category_name')[col]
        .transform(lambda x: x.fillna(x.median()))
    )
    products[col] = products[col].fillna(products[col].median())
print('\n\nAfter')
print(products.shape)
print(products.isna().sum())
```

```python
# sellers
print('Before')
print(sellers.isna().sum())
print('shape = ',sellers.shape)

sellers = sellers.merge(geo_location , how='left' , left_on='seller_zip_code_prefix' , right_on='geolocation_zip_code_prefix')
sellers['seller_city'] =sellers['seller_city'].fillna(sellers['geolocation_city'])
sellers['seller_state'] =sellers['seller_state'].fillna(sellers['geolocation_state'])
sellers = sellers.drop(columns=[
    'geolocation_zip_code_prefix',
    'geolocation_lat',
    'geolocation_lng',
    'geolocation_city', 
    'geolocation_state']
)
sellers.isna().sum()


print('\n\nAfter')
print(sellers.isna().sum())
print('shape = ',sellers.shape)
```
